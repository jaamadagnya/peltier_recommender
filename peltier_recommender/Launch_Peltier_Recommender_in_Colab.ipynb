This notebook launches the bundled Streamlit decision-support application. You need the supplied `peltier_recommender.zip` file and a free [ngrok account](https://dashboard.ngrok.com/signup) with an authentication token.

```python
#@title 3. Enter your ngrok authentication token
from getpass import getpass
from pyngrok import ngrok

token = getpass("Paste your ngrok authentication token: ")
ngrok.set_auth_token(token)
print("Token configured for this temporary Colab session.")
```

```python
#@title 4. Launch the recommendation system
import subprocess, time
from pyngrok import ngrok

ngrok.kill()
process = subprocess.Popen(
    ["streamlit", "run", "peltier_recommender/app.py",
     "--server.port", "8501", "--server.headless", "true"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.STDOUT,
)
time.sleep(5)
tunnel = ngrok.connect(8501)
print("Open the recommendation system here:")
print(tunnel.public_url)
```

## Use the tool

1. Select a preset or choose custom priorities.
2. Optionally activate engineering limits for any of the six responses.
3. Review the recommended configuration, runner-up, margin and explanation.
4. If the application reports an indifference region, treat both leading configurations as practically equivalent under the selected priorities.
5. Download the recommendation record as CSV.

Recommendations are limited to the 48 experimentally evaluated configurations and should not be interpreted as predictions for untested designs.
